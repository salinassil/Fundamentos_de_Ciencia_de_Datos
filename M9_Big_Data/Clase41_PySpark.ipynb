```python
from pyspark import SparkContext, SparkConf
```

```python
conf = SparkConf().setAppName("SilApp").setMaster("local[*]")
sc = SparkContext.getOrCreate(conf)

print(sc)
```

```python
lista=[1,2,3,4,5]
```

```python
rdd_0=sc.parallelize(lista)
```

```python
rdd_0
```

```python
resultado=rdd_0.filter(lambda x: x%2==0)
```

```python
print(resultado)
```

```python
print(resultado.collect())
```
