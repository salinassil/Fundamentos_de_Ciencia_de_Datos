```python
from pyspark.sql import DataFrame, SparkSession
```

```python
spark_builder=SparkSession.builder.appName('Practice').getOrCreate()
spark_builder
```

```python
data = [
        ('John','Frererich','Smith','1976-05-05','M',11003.98),
        ('Jane','Riana','Smith','1984-01-27','F',123324.90),
        ('Jonas','Jason','Smith','1989-08-02','M',133324.50),
]
columns = ['first_name','middle_name','last_name','date_of_birth','gender','salary']
```

```python
df=spark_builder.createDataFrame(data=data,schema=columns)
df.show()
```

```python
df
```

```python
df.createOrReplaceTempView("df_prueba") #crea una tabla temporal explorable con SQL
```

```python
genero=spark_builder.sql("SELECT * FROM df_prueba WHERE salary > 120000")
genero.show()
```

```python
import requests
path = "https://raw.githubusercontent.com/owid/covid-19-data/master/public/data/owid-covid-data.csv"
req = requests.get(path)
url_content = req.content

csv_file_name = 'owid-covid-data.csv'
csv_file = open(csv_file_name, 'wb')

csv_file.write(url_content)
csv_file.close()
```

```python
csv_file
```

```python
df_csv = spark_builder.read.csv('/content/'+csv_file_name, header=True, inferSchema=True)
df_csv.show()
```
